```python
# 1.remove duplicate elements

lst = [1, 2, 2, 3, 4, 4, 5]
res = []
for i in lst:
    if i not in res:
        res.append(i)
print(res)
```

```python
# 2.find second largest number

lst=[10, 5, 8, 20, 15]
max_val=lst[0]
for i in lst:
    if i>max_val:
        max_val = i
sec_max = lst[0]
for i in lst:
    if i!=max_val and i>sec_max:
        sec_max = i
print(sec_max)
```

```python
# 3.count character frequency

s = input("enter string: ")
freq={}
for ch in s:
    if ch in freq:
        freq[ch]+=1
    else:
        freq[ch]=1
print(freq)
```

```python
# 4.print prime numbers

start = int(input())
end = int(input())
for n in range(start, end + 1):
    if n > 1:
        f = 0
        for i in range(2, n):
            if n % i == 0:
                f = 1
                break
        if f == 0:
            print(n)
```

```python
# 5.fibonacci series

def fib(n):
    a, b = 0, 1
    for _ in range(n):
        print(a)
        a, b = b, a + b

n = int(input())
fib(n)
```

```python
# 6.reverse each word

s=input()
words=s.split()
ans=""
for w in words:
    temp=""
    for i in range(len(w)):
        temp=w[i]+temp
    ans = ans+temp+" "
print(ans)
```

```python
# 7.count characters

s=input()
a=0
d=0
sp=0
for ch in s:
    if ch.isalpha():
        a=a+1
    elif ch.isdigit():
        d=d+1
    else:
        sp=sp+1
print(a,d,sp)
```

```python
# 8.find common elements

l1=[1, 2, 3, 4, 5]
l2=[4, 5, 6, 7, 8]
c=[]
for i in l1:
    if i in l2 and i not in c:
        c.append(i)
print(c)
```

```python
# 9.list operations

lst = [10, 20, 30]
lst.append(40)
lst.remove(20)
print(max(lst))
print(min(lst))
print(len(lst))
```
