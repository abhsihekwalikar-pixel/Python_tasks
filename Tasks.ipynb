```python
# 2 Basic arithmetic operations

x = float(input("Enter first number:"))
y = float(input("Enter second number:"))
print("Addition:",x+y)
print("Subtraction:",x-y)
print("Multiplication:",x*y)
print("Division:",x/y)
```

```python
# 3 Check even or odd

A = int(input("Enter a number: "))
if A % 2 == 0:
    print(A, "is Even")
else:
    print(A, "is Odd")
```

```python
# 4 Largest of three numbers

x=float(input("first number:"))
y=float(input("second number:"))
z=float(input("third number:"))
if x>=y and x>=z:
    print("Largest:",x)
elif y>=x and y>=z:
    print("Largest:",y)
else:
    print("Largest:",z)
```

```python
# 5 Print 1 to 100 using loops
print("For Loop:")
for i in range(1, 101):
    print(i)

print("\nWhile Loop:")
n = 1
while n <= 100:
    print(n)
    n += 1
```

```python
# 6 Calculate factorial

num = int(input("Enter a number:"))
if num < 0:
    print("Factorial not defined for negative numbers")
elif num == 0:
    print("Factorial is 1")
else:
    fact = 1
    for i in range(1, num + 1):
        fact = fact * i
    print("Factorial:", fact)
```

```python
# 7. List operations

m = [34,12,56,78,23,90,45,67,11,89]
print("Original:", m)
m.append(100)
print("After append:", m)
m.remove(56)
print("After remove:", m)
m.sort()
print("After sort:", m)
m.reverse()
print("After reverse:", m)
```

```python
# 8 Count vowels in string

s = input("Enter a string:").lower()
count = 0
for ch in s:
    if ch in "aeiou":
        count = count + 1
print("Total vowels:", count)
```

```python
# 9 Student dictionary and print values
stu = {"Name": input("Name:"),
    "Age": int(input("Age:")),
    "Course": input("Course:")}
print("Student details:")
for k in stu:
    print(k, ":", stu[k])
```

```python
# 10 Check palindrome

txt = input("Enter a string:")
rev = txt[::-1]
if txt == rev:
    print("It is a palindrome")
else:
    print("Not a palindrome")
```
